```python
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
import pandas as pd
df=pd.read_csv('sales_data.csv')

df['Revenue']=df['Price']*df['Quantity']
df.head()
```

```python
df.tail()
```

```python
## Basic Scatter plot

import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("scatter plot of total bills vs tip")
plt.show()
```

```python
##Line plot

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title('lineplot of size vs total bill')
plt.show()
```

```python
#Bar plots

sns.barplot(x='day',y='total_bill',data=tips)
plt.title('Barplot of day and total bill')
plt.show()
```

```python
#Box plots

sns.boxplot(x='day',y='total_bill',data=tips)
plt.title("Box plot of day and totalbill")
plt.show()
```

```python
#Violin plot

sns.violinplot(x='day',y='total_bill',data=tips)
plt.title("Violin plot of day and total bill")
plt.show()
```

```python
#Histplot

sns.histplot(tips['total_bill'],bins=10,kde=True)      #if kde is false then curve would dissappear
```

```python
#kde plot

sns.kdeplot(tips['total_bill'],fill=True)
```

```python
#Comparing the plots

sns.pairplot(tips)
```

```python
#Correlation between the numerical columuns it ranges from -1 to 1 

corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
#Heatmap

sns.heatmap(corr,annot=True,cmap='coolwarm')
```
