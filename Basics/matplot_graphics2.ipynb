Matplotlib is the foundation for plotting in python , while seaborn is a higher-level library built on top of matplotlib , which makes visualisations easier and prettier

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("student_report.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
plt.figure(figsize=(8,5))

#s is scatter size and hue defines where the data point belongs to

sns.scatterplot(data=df,x='Attendance',y='Assignment_Score',hue='Result',s=100)

plt.title("Result based on the attendance and assignment_score")
plt.xlabel("Attendance")
plt.ylabel("Assignment score")


plt.show()
```

```python
sns.countplot(data=df, x='Result')

plt.title("Pass vs fail")
plt.xlabel("Result")
plt.ylabel("Students")
plt.show()
```

```python
sns.stripplot(
    data=df,
    x='Result',
    y='Attendance',
    jitter=True,
    s=8
)

plt.title("individual attendence by result")
```

```python
sns.barplot(
    data=df,
    x='Study_Hours',
    y='Result',
)
plt.title("Barplot of students")

```

```python
sns.kdeplot(
    data=df,
    x='Study_Hours',
    hue='Result',
    fill=True
)

plt.title("Curve plotting")
```

```python
sns.pairplot(df,hue='Result')   #Gives the relationship between all features
plt.show()
```

```python
corr=df.drop(columns='Result').corr()
corr
```

```python
sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)
plt.title("Correlation heatmap")
plt.show()
```

```python
sns.regplot(
    data=df,
    x='Sleep_Hours',
    y='Assignment_Score'
)
plt.title("Regression plot of sleep vs score")
```

```python
#Scatter plot + distribution of both variables

sns.jointplot(
    data=df,
    x='Sleep_Hours',
    y='Assignment_Score',
    hue='Result'
)
```

```python
sns.swarmplot(
    data=df,
    x='Result',
    y='Assignment_Score'
)
```

```python
sns.displot(
    data=df,
    x='Sleep_Hours',
    hue='Result'
)
```
