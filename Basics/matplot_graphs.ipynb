```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_wine
wine=load_wine()
print(wine.DESCR)       #gives info , attributes and instances
```

```python
wine_data=pd.DataFrame(wine.data)
wine_data.columns=wine.feature_names
wine_data.head()
```

```python
wine_data.tail()
```

```python
plt.scatter(wine_data['malic_acid'],wine_data['alcohol'])
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=wine_data,
    x='alcohol',
    bins=20,
    kde=True        #kde is the curve
)
plt.title("Alcohol distribution")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```

```python
wine_data['target']=wine.target
wine_data['target_name']=wine_data['target'].map({
    0:'Class 0',
    1:'Class 1',
    2:'Class 2'
})

plt.figure(figsize=(8,5))

sns.scatterplot(data=wine_data,x='alcohol',y='malic_acid',hue='target_name',s=100)  #s is scatter size and hue defines where the data point belongs to
plt.title("Malic acid vs alcohol")
plt.xlabel("alcohol")
plt.ylabel("malic_acid")
plt.show()
```
