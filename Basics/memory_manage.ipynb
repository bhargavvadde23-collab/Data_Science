Refcount

Refcount is a count of objects pointing to that exact variable
when the variable is deleted python dynamically deallocates the memory

```python
import sys

a=[]

print(sys.getrefcount(a))
```

```python
b=a
print(sys.getrefcount(b))
```

```python
import gc

gc.collect()    #the number of unreachable objects are created

print(gc.get_stats())
print(gc.garbage)
```
