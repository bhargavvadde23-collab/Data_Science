Process

Process is simply the program under execution or that is being executed

Thread 

Thread is a unit of execution within a process

A Process can contain data segment,header,heap,  so the thread uses those segments and it has it's own 
Stack and register

Difference between threads and process

Threads allows concurrent execution , like it allows the program to run two functions concurrently 
with the shared memory 

Process allows parallel execution , like it allows  program to run two functions parallely 
with separate memory
When we execute the process program , it creates the separate processes for each process

```python
import threading
import time

def print_numbers():
    for i in range(5):
        time.sleep(2)
        print(f"Numbers:{i}")
        
def print_letters():
    for letter in "abcde":
        time.sleep(2)
        print(f"Letters:{letter}")
        
t=time.time()

finished_time=time.time()-t

t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_letters)

t1.start()
t2.start()

t1.join()
t2.join()



# if we call the functions without threading concept the time takes for this program
#is 20 sec so there is no concurrent execution of those functions
#But with the threads we can execute the two functions concurrently, so time taken is half normal calls

#print_numbers()
#print_letters() 

print(f"Finished time:{finished_time}")
        
```

```python
import multiprocessing
import time

def squares():
    for i in range(5):
        time.sleep(2)
        print(f"Squares:{i*i}")
        
def cubes():
    for i in range(5):
        time.sleep(2)
        print(f"Cubes:{i*i*i}")
        
        
if __name__=="__main__":
    
    #create process
    
    p1=multiprocessing.Process(target=squares)
    p2=multiprocessing.Process(target=cubes)
    
       
    t=time.time()
    
    p1.start()
    p2.start()
    
    p1.join()
    p2.join()
    
    
    finished_time=time.time()-t
    print(finished_time) 
        
```

Multi threading is used for i/o bound tasks

Scenario : Web Scrapping (Fetching data from web pages)

Web scrapping involves making numerous requests from web pages , this tasks are i/o bound because 
they spend a lot time wating for response from servers , multi threading improves performance by allowing
multiple web pages fetched concurrently

```python
from bs4 import BeautifulSoup
import threading
import requests


'''
https://en.wikipedia.org/wiki/LangChain

https://en.wikipedia.org/wiki/Microsoft

https://en.wikipedia.org/wiki/Amazon_(company)
'''

urls=[
'https://en.wikipedia.org/wiki/LangChain',

'https://en.wikipedia.org/wiki/Microsoft',

'https://en.wikipedia.org/wiki/Amazon_(company)'
]


def fetch_content(url):
    response=requests.get(url)
    soup=BeautifulSoup(response.content,'html.parser')
    print(f"Fetched {len(soup.text)} characters form {url}")
          

threads=[]

for url in urls:
    thread=threading.Thread(target=fetch_content,args=(url,))
    threads.append(thread)
    thread.start()
    
for thread in threads:
    thread.join()
    
print("All webpages fetched succesfu")
```
