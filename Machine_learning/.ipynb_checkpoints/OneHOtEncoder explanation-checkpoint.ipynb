```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
```

```python
df=pd.read_csv('gender_city.csv')
df.reset_index(drop=True)
df.drop(['ID'],axis=1,inplace=True)
```

```python
### Sparse output is a parameter used in the onehot encoder 
Mostly it consists of 0's , so to store only the ones , default it creates the matrix of 1. ignores the 0's
```

```python
encoder=OneHotEncoder(sparse_output=False)
label=LabelEncoder()

df['Gender']=label.fit_transform(df['Gender'])
```

```python
df
```

```python
encoded_data=encoder.fit_transform(df[['City']])
```

```python
encoded_data
```

```python
encoded_df=pd.DataFrame(encoded_data,columns=encoder.get_feature_names_out(['City']))
encoded_df
```

```python
df=pd.concat([df,encoded_df],axis=1)
```
