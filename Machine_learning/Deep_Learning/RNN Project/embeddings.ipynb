### One hot 
One_Hot in tensorflow is used to convert sentences and words into numerical vectors
They usually return in sparse vectors
So that's why we use Embeddings
Embeddings return the sparse matrix into dense matrix , some meaningful information

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent= [
    'this is milk',
    'i am a boy',
    'He is playing',
    'you will never know the pain',
    'Dog is playing outside',
    'Welcome to the world'
]
```

```python
sent
```

```python
voc_size=100
```

### It gives the indices to the words in the sentences

```python
one_hot_repr=[one_hot(words,voc_size) for words in sent]
one_hot_repr
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
sent_length=8
embeded_docs=pad_sequences(one_hot_repr,maxlen=sent_length,padding='pre')
embeded_docs
```

```python
dim=10
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embeded_docs)
```
