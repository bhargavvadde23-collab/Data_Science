### Bow and Tf-idf machine learning algorithms or techniques
In general Machine Computers do not understands the direct sentences , so the sentences must convert into numerical values .
So we use the techniques called Bag of words and term Frequeny - invert document frequency 
Each has differrent approaches 
1)Bag of Words
It uses the vocabulary(unique words)
and then calculates the value for the words , It gives equal priority to every word

2)Term Frequency and Invert Document Frequency
It gives high priority to the important words
It often measures how many times a word appeared in the sentece
and applies a logarithmic Formula

### Word2Vec 
    -> It is also a nlp technique to convert words to vectors 
    -> It was published by the google 
    -> The model trains over a 3 billion words
    -> The words are converted in to vectors , intially the model assigns random values to the words , each word represented as the 
    -> 300 dimensional vector and lastly averages the vector values into a single vector of 300 dimension

### Basic Workflow of a machine learning algorithm
    ->Preprocessing and Cleaning
    ->Train and Test split
    ->BOW and TF-IDF (Sentences --> Vectors)
    ->ML Models To train the data


### Re Module
re stands for Regular Expressions is a Module in python is used in searching ,matching , replacing , and manipulating the text patterns
-> To keep only alphabets in the dataset 
        -> use text='i love! python, programming language'
        -> clean_text=re.sub(r'[^a-zA-Z ]','',text)
    -> re.sub(pattern,replacement,string)

### Split function returns a list of words from a text
    text=text.split() -> default method to separate the text based on the space
    text=text.split(',') -> separate based on the comma

### Difference betweeen the fit_transform and Transform
    fit-> learns patterns from the data like a relationship 
    transform-> applies that pattern to the data
So we use the fit_transform on the train data to learn a relationship
and apply the transform on the test data on unseen data like applying a same relationship on the unseen data

### Cosine similarity 
        -> It is a measure used to determine how similar two vectors are
        -> It tells is the two values related or not
        -> it ranges from -1 0 1

```python
import pandas as pd
df=pd.read_csv("SMSSpamCollection",sep='\t',names=['label','message'])
```

```python
df.head()
```

### Make a Small dataset from the big dataset
df.sample --> used random samples
df.head(150) --> used first 150 samples

```python
df_small=df.sample(n=150,random_state=42)
df_small=df_small.reset_index(drop=True)
```

```python
df_small.head()
```

```python
df_small
```

```python
df_small['label'].value_counts()
```

```python
df_small.shape
```

```python
df_small['message']
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
lemmatizer=WordNetLemmatizer()
import re
```

```python
corpus=[]
for i in range (0,len(df_small)):
    review=re.sub('[^a-zA-Z]',' ',df_small['message'][i])
    review=review.lower()
    review=review.split()
    review=[lemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
### Extract Output feature
y=pd.get_dummies(df_small['label'])
```

```python
y
```

```python
y=y.iloc[:,0].values
```

```python
y
```

```python
type(corpus),type(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.2)
```

```python
len(X_train),len(y_train)
```

```python
len(X_test),len(y_test)
```

```python
### For small dataset we don't use the max_feature parameter
#Bag of words
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer(ngram_range=(1,2))
```

```python
X_train=cv.fit_transform(X_train).toarray()
X_test=cv.transform(X_test).toarray()
```

```python
X_train
```

```python
cv.vocabulary_
```

## Machine Learning Model Multinomial naive bayes

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_model_bow=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred_bow=spam_model_bow.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report

print(accuracy_score(y_test,y_pred_bow))
print(classification_report(y_test,y_pred_bow))
```

## Tf-idf

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(corpus,y,test_size=0.2)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tv=TfidfVectorizer(ngram_range=(1,2))

X_train=tv.fit_transform(X_train).toarray()
X_test=tv.transform(X_test).toarray()
```

```python
X_train
```

```python
X_test
```

```python
cv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB

spam_tfidf_model=MultinomialNB().fit(X_train,y_train)
```

```python
y_pred_tfidf=spam_tfidf_model.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred_tfidf))
```

```python
print(classification_report(y_test,y_pred_tfidf))
```
