### Lemmatization 
Lemmatization is a technique like stemming , the output of the lemmatization is called lemma,which is a root word not stem word
After Lemmatization we'll get the valid word as the output , we need to specify the pos(Parts of speech)

```python
from nltk.stem import WordNetLemmatizer
```

```python
import nltk
nltk.download('wordnet')
```

```python
lemma=WordNetLemmatizer()
```

```python
lemma.lemmatize("going",pos='v')
```

```python
lemma.lemmatize("congratulations",pos='a')
```

```python
'''
Noun-n
verb-v
adverb-a
adverb-r
'''
```

```python
lemma.lemmatize('going',pos='v')
```

```python
lemma.lemmatize('fairly'),lemma.lemmatize('sportingly')
```

```python
lemma.lemmatize('willingly',pos='v')
```

```python
words=['programming','programs','finalise','final','greetings','congratulate']
```

```python
for word in words:
    print(word+"---->"+lemma.lemmatize(word))
```
