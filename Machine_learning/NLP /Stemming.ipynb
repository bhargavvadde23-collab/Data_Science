### Stemming
Stemming is the process of reducing the word to its word stem
Stemming is important in natural language understanding (NLU) and natural language processing (NLP)

### Lemmatization 
Lemmatization is the process of reducing the word to its root word , parts of speech can also be used in the syntax.

### Porter Stemmer
It may not give the accurate results
But Snowball Stemming may give the accurate results

```python
from nltk.stem import PorterStemmer
```

```python
words=['eating','eats','sleeping','sleep','finally','finalising','congratulations','congratulating',
      'programming','programs']
```

```python
porter=PorterStemmer()
```

```python
porter.stem("congratulate")
```

```python
porter.stem('finalize')
```

```python
porter.stem('hypnotizing')
```

```python
for word in words:
    print(word+"---->"+porter.stem(word))
```

```python
from nltk.stem import SnowballStemmer
```

```python
snow=SnowballStemmer('english')
```

```python
for word in words:
    print(word+'---->'+snow.stem(word))
```

### It is still better than porter stemming , thats why it is called as porter2

```python
porter.stem('fairly'),porter.stem('sportingly')
```

```python
snow.stem('fairly'),snow.stem('sportingly')
```
