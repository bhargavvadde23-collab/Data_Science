Decision tree is a supervised machine learning which makes  predictions by asking a series of questions
about the input data.

For continuous value DecisonTreeRegressor
For classification value DecisionTreeClassifier

```python
from sklearn.datasets import load_diabetes
```

```python
df1=load_diabetes()
```

```python
df1
```

```python
print(df1['DESCR'])
```

```python
import pandas as pd
```

```python
df2=pd.DataFrame(df1.data,columns=
                 ['age','sex','bmi','bp','s1','s2','s3','s4','s5','s6'])
```

```python
df2
```

```python
df2.head()
```

Actually target value contains 1 year diabetes disease progress

```python
df1['target']
```

```python
X=df2
y=df1['target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

```python
y_train
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
regressor=DecisionTreeRegressor(max_depth=4)
```

```python
regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
print(r2_score(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```

```python
from sklearn import tree
import matplotlib.pyplot as plt
plt.figure(figsize=(15,8))
tree.plot_tree(regressor,filled=True)
```
