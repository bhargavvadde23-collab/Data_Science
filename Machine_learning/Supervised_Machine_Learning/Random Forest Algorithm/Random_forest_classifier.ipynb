Many machine learning models works on only numerical data not text data , if the dataset contains city,marriage status,etc... it is in text form so we use onehot encoder to convert the textual classifier values to numerical boolean values , So that the model will be trained and tested on numerical data

If we are using any model , the dataset contains numerical values and the categorical values , so we scale the numerical vallues and encode the categorical values , So both have to do manually and join after doing them , the code looks messy , To overcome this difficulty we use ColumnTransformers , it would scale the numerical data and onehot the categorical data and combines them using a single variable . so it woould be easy to train and test the training and testing the data ...
Remember --> only use fit_transform on training data ad use transform on test data

## Imputing the null values
### Age with median
### Typeofcontact with mode
### Duration of pitch with median
### Numberoffollowups with mode
### Preferredpropertystar with mode
### NumberofTrips with median
### Numberofchildrevisiting with mode
### Monthly income with median


```python
df.Age.fillna(df.Age.median(),inplace=True)
df.TypeofContact.fillna(df.TypeofContact.mode()[0],inplace=True)
df.DurationOfPitch.fillna(df.DurationOfPitch.median(),inplace=True)
df.NumberOfFollowups.fillna(df.NumberOfFollowups.mode()[0],inplace=True)
df.PreferredPropertyStar.fillna(df.PreferredPropertyStar.mode()[0],inplace=True)
df.NumberOfTrips.fillna(df.NumberOfTrips.median(),inplace=True)
df.NumberOfChildrenVisiting.fillna(df.NumberOfChildrenVisiting.mode()[0],inplace=True)
df.MonthlyIncome.fillna(df.MonthlyIncome.median(),inplace=True)
```

```python
df.isnull().sum()
```

```python
df.drop(columns='CustomerID',inplace=True,axis=1)
```

```python
df.head()
```

```python
df['TotalVisiting']=df['NumberOfPersonVisiting']+df['NumberOfChildrenVisiting']
df.drop(columns=['NumberOfPersonVisiting','NumberOfChildrenVisiting'],inplace=True,axis=1)
```

## Feature Ectraction

Standardizing the numerical data and onehot encoding the categorical data

```python
#Numerical Features
num_features=[features for features in df.columns if df[features].dtype != 'O']
print("No.of numerical columns is :",len(num_features))
```

```python
cat_features=[features for features in df.columns if df[features].dtype=='O']
print("no.of of categorical features is ", len(cat_features))
```

```python
from sklearn.model_selection import train_test_split
X=df.drop(columns='ProdTaken',axis=1)
y=df['ProdTaken']
```

```python
X.head()
```

```python
y
```

```python
y.value_counts()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
X_train.shape,y_train.shape
```

```python
X.info()
```

```python
cat_features=X.select_dtypes(include='object').columns
num_features=X.select_dtypes(exclude='object').columns

from sklearn.preprocessing import OneHotEncoder , StandardScaler
from sklearn.compose import ColumnTransformer

num_scaler = StandardScaler()
oh_transformer = OneHotEncoder(drop='first')

preprocesor= ColumnTransformer(
    [
        ('OneHotEncoder',oh_transformer,cat_features),
        ('StandardScaler',num_scaler,num_features)
    ]
)
```

```python
preprocesor
```

```python
X_train.head()
```

```python
X_train=preprocesor.fit_transform(X_train)
```

```python
pd.DataFrame(X_train)
```

```python
X_test=preprocesor.transform(X_test)
```

```python
X_test
```

```python
y_train
```

```python
y_train.value_counts()
```

### Random Forest Classifier
Training and testing on Different Models 

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import precision_score,recall_score,f1_score,accuracy_score,roc_auc_score
```

```python

models={
    'Logistic Regression':LogisticRegression(),
    'DecisionTree Classifier':DecisionTreeClassifier(),
    'RandomForest Classifier':RandomForestClassifier(),
    'GradientBoost Classifier':GradientBoostingClassifier()
}

for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(X_train,y_train)
    
    #Make predictions
    y_pred_train=model.predict(X_train)
    y_pred_test=model.predict(X_test)
    
    
    #Training score
    model_train_accuracy=accuracy_score(y_train,y_pred_train)
    model_train_recall_score=recall_score(y_train,y_pred_train)
    model_train_f1_score=f1_score(y_train,y_pred_train , average='weighted')
    model_train_precision_score=precision_score(y_train,y_pred_train)
    model_train_roc_auc_score=roc_auc_score(y_train,y_pred_train)
    
    #Test score
    model_test_accuracy=accuracy_score(y_test,y_pred_test)
    model_test_recall_score=recall_score(y_test,y_pred_test)
    model_test_f1_score=f1_score(y_test,y_pred_test , average='weighted')
    model_test_precision_score=precision_score(y_test,y_pred_test)
    model_test_roc_auc_score=roc_auc_score(y_test,y_pred_test)
    
    
    print(list(models.keys())[i])
    
    print("Model performance on training data")
    
    print("-Accuracy :{:.4f}".format(model_train_accuracy))
    print("-recall_score :{:.4f}".format(model_train_recall_score))
    print("-f1_score :{:.4f}".format(model_train_f1_score))
    print("-Precision_score :{:.4f}".format(model_train_precision_score))
    print("-roc_auc_score :{:.4f}".format(model_train_roc_auc_score))
    
    
    print("------------------------------------------")
    
    print("Model performance on testing data")
    
    print("-Accuracy :{:.4f}".format(model_test_accuracy))
    print("-recall_score :{:.4f}".format(model_test_recall_score))
    print("-f1_score :{:.4f}".format(model_test_f1_score))
    print("-Precision_score :{:.4f}".format(model_test_precision_score))
    print("-roc_auc_score :{:.4f}".format(model_test_roc_auc_score))
    


    print("="*35)
    print("\n")
    
    
    
    
    
    
```
