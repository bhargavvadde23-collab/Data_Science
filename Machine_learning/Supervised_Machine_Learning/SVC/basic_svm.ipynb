```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.svm import SVC

# 1. Generate a simple 2D linearly separable dataset
X, y = make_blobs(n_samples=100, centers=2, random_state=42, cluster_std=0.60)

# 2. Initialize and train the SVM model with a linear kernel
model = SVC(kernel='linear', C=1.0)
model.fit(X, y)

# 3. Set up the plot
plt.figure(figsize=(8, 6))
ax = plt.gca()

# Scatter plot of the data points
plt.scatter(
    X[:, 0], X[:, 1], c=y, s=50, cmap='autumn', edgecolors='k', label='Data Points'
)

# Get limits of the current plot
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# 4. Create a grid (mesh) to evaluate the decision function
xx = np.linspace(xlim[0], xlim[1], 30)
yy = np.linspace(ylim[0], ylim[1], 30)
YY, XX = np.meshgrid(yy, xx)
xy = np.vstack([XX.ravel(), YY.ravel()]).T
Z = model.decision_function(xy).reshape(XX.shape)

# 5. Plot the decision boundary and margins
# levels=[-1, 0, 1] plots the lower margin, hyperplane (0), and upper margin
ax.contour(
    XX,
    YY,
    Z,
    colors='k',
    levels=[-1, 0, 1],
    alpha=0.6,
    linestyles=['--', '-', '--'],
)

# 6. Highlight the Support Vectors
ax.scatter(
    model.support_vectors_[:, 0],
    model.support_vectors_[:, 1],
    s=150,
    linewidth=1.5,
    facecolors='none',
    edgecolors='black',
    label='Support Vectors',
)

# 7. Final plot adjustments
plt.title('SVM Decision Boundary, Margins, and Support Vectors')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import plotly.express as px

# 1. Load sample dataset (Gapminder data)
df = px.data.gapminder()
df_2007 = df.query("year == 2007")

# 2. Create an interactive scatter plot
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    log_x=True,
    size_max=60,
    title="<b>GDP vs Life Expectancy (2007)</b>",
    labels={
        "gdpPercap": "GDP per Capita (USD)",
        "lifeExp": "Life Expectancy (Years)",
    },
)

# 3. Display the interactive plot
fig.show()
```
