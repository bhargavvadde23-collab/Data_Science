#### DBSCAN (Density based spatial clustering of application with noise)
DBSCAN is a popular algorithm majorly used for detecting the outliers , it contains two parameters called
eps= maximum distance between two points for them tobe considered as neighbors
and min_samples=5 , 5 points are considered as the core points within the eps radius value

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_circles
```

Here y is the dependent variable we do not use it , as we are dealing with the unsupervised machine learning

```python
X,y=make_circles(n_samples=700,factor=0.3,noise=0.1)
```

```python
X
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbscan=DBSCAN(eps=0.1)
dbscan.fit_predict(X)
```

```python
labels=dbscan.fit_predict(X)
outliers=X[labels==-1]
```

```python
outliers
```

### Only the outliers

```python
plt.scatter(X[:,0],X[:,1])
plt.scatter(outliers[:,0],outliers[:,1])
```
