### IsolationForest 
Isolation Forest is Machine Learning algorithm which is used to detect the outliers based on the threshold value . 
It contains a parameter called " contamination ", Tells the percentage of outliers present in the dataset
We have to initialize it like 0.02 or 0.5 or 0.2 , based on how much outliers present in the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("health_care.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df.shape
```

Analyze the outliers

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1])
```

```python
from sklearn.ensemble import IsolationForest
```

```python
isf=IsolationForest(contamination=0.2)
isf.fit(df)
predictions=isf.predict(df)
```

```python
predictions
```

```python
print(type(predictions))
```

```python
anamoly_index=np.where(predictions < 0)
anamoly_index
```

```python
x=df.values
```

```python
x
```

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[anamoly_index,0],x[anamoly_index,1])
```

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[anamoly_index,0],x[anamoly_index,1],edgecolor='r')
```

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[anamoly_index,0],x[anamoly_index,1],color='g')
```
