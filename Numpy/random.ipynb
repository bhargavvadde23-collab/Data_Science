### np.where is a function which returns the list of indecies where the certain condition is met

It is a useful function

```python
arr=np.array([10,20,30,40,50])
result=np.where(arr > 20)
print(result)               #returns the indecies
```

```python
arr=np.array([10,20,30,40,50])
result=np.where(arr > 20 , "High","low")
print(result)
```

```python
arr=np.array([10,20,30,40,50])
result=np.where(arr > 20 ,1,0)
print(result)
```
